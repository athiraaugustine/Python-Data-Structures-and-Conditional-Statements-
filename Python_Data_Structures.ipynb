#### SET

```python
my_set={'a','e','i','o','u','a','a','i'}
my_set        
```

'''
- A set automatically removes duplicates value and store only unique elements. 
- The output order may vary because sets are unorderd.
'''

```python
my_set={'a','e','i','o','u','a','a','i'}
my_set[4]='s'
```

'''
- Sets are unorderd collections.
- Sets do not support indexing.
- Individual elements cannot be modified using index assignment.
- Use add() to insert a new element into a set.
'''

```python
my_set.add('s')
print(my_set)      # this method is used to insert new element in the set.
```

```python
set1 ={1.3,5,7,9}
set2={2,3,5,8,10}
print(set1)
print(set2)
```

```python
set1 ={1,3,5,7,9}
set2={2,3,5,8,10}
print('union:',set1.union(set2))
```

```python
set1 ={1,3,5,7,9}
set2={2,3,5,8,10}
print('intersection:',set1.intersection(set2))
```

#### OPERATORS / CONDITIONAL STATEMENTS

```python
score=int(input('enter your score (0 to 10) :'))

if 7 < score <= 10:
    print('Above Average : Great job! Keep up the good work.')
elif 4 <= score <= 7:
    print('Average : Good effort! A little more practice will make a difference.')
elif 0 <= score < 4:
    print('Below Average : Keep trying! Improvement comes with consistent effort.')
else:
    print('Invalid score. Please enter score between 0 and 10')
```
